```python
import pandas as pd
import numpy as np
from nba_api.stats.endpoints import leaguedashplayerstats
```

```python
#get league and player stats for the 2024-25 NBA season via the nba_api library and convert to dataframe
player_stats = leaguedashplayerstats.LeagueDashPlayerStats(season = '2024-25').get_data_frames()[0]
player_stats.head()
```

```python
#identify column headers
print(f"Column Headers:")
print("-"*25)
player_stats.columns.tolist()
```

# Transformation 1: Standardize Player Names

## Create a new column that shows player name cleaned up to all lower letters and remove any punctuation

```python
player_stats['Player_clean'] = player_stats['PLAYER_NAME'].str.lower().str.strip().replace(',', '').str.replace('.', '')
player_stats.head()
```

# Transformation 2: Create a new column for whole team name from abbreviation

```python
#Step 1: get the abbreviated team names used in the dataset
unique_values = player_stats['TEAM_ABBREVIATION'].unique()
unique_values
```

```python
#Step 2: created a dictionary for the full team names, ignoring nan
team_full_name = {
    'ATL': 'Atlanta Hawks',
    'BOS': 'Boston Celtics',
    'BKN': 'Brooklyn Nets',
    'CHA': 'Charlotte Hornets',
    'CHI': 'Chicago Bulls',
    'CLE': 'Cleveland Cavaliers',
    'DAL': 'Dallas Mavericks',
    'DEN': 'Denver Nuggets',
    'DET': 'Detroit Pistons',
    'GSW': 'Golden State Warriors',
    'HOU': 'Houston Rockets',
    'IND': 'Indiana Pacers',
    'LAC': 'Los Angeles Clippers',
    'LAL': 'Los Angeles Lakers',
    'MEM': 'Memphis Grizzlies',
    'MIA': 'Miami Heat',
    'MIL': 'Milwaukee Bucks',
    'MIN': 'Minnesota Timberwolves',
    'NOP': 'New Orleans Pelicans',
    'NYK': 'New York Knicks',
    'OKC': 'Oklahoma City Thunder',
    'ORL': 'Orlando Magic',
    'PHI': 'Philadelphia 76ers',
    'PHX': 'Phoenix Suns',
    'POR': 'Portland Trail Blazers',
    'SAC': 'Sacramento Kings',
    'SAS': 'San Antonio Spurs',
    'TOR': 'Toronto Raptors',
    'UTA': 'Utah Jazz',
    'WAS': 'Washington Wizards'
}

#Step 3: create new column with team full names from dictionary
player_stats['Whole Team Name'] = player_stats['TEAM_ABBREVIATION'].map(team_full_name)
player_stats.head()
```

```python
#validate the changes in the new col
unique_values2 = player_stats['Whole Team Name'].unique()
unique_values2
```

# Transformation 3: Update percentage column entries from decimal format to more readable percentage format

```python
player_stats['W_PCT'] = player_stats['W_PCT'] * 100
player_stats['FG_PCT'] = player_stats['FG_PCT'] * 100
player_stats['FG3_PCT'] = player_stats['FG3_PCT'] * 100
player_stats['FT_PCT'] = player_stats['FT_PCT'] * 100

player_stats.head()
```

# Transformation 4: Change 'MIN' column name

## I kept reading 'MIN' as minimum, so updating column heading to be more understandable for Minutes Played.

```python
player_stats = player_stats.rename(columns = {'MIN': 'Mins_Played'})
player_stats.head()
```

# Transformation 5: Remove decimals from Mins_Played

## Since minutes played currently has six decimal places, rounding up to make more readable.

```python
player_stats['Mins_Played'] = np.ceil(player_stats['Mins_Played'])
player_stats.head()
```

# Transformation 6: Check for any duplicate rows

```python
#Step 1: determine the amount of duplicate rows
duplicates = player_stats[player_stats.duplicated(keep=False)]
duplicates
```

Since no duplicates found, no further action needed.
